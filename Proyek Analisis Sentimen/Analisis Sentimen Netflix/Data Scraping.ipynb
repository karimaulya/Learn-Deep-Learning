# import data review

```python
from google_play_scraper import Sort, reviews_all, reviews
import pandas as pd

app_id = "com.netflix.mediaclient"

result, continuation_token = reviews(
    app_id,
    lang = "id",
    country = "id",
    sort = Sort.NEWEST,
    count = 11000,
)

df = pd.DataFrame(result)
print(f"Berhasil scraping {len(df)} review")
df[["userName", "score", "content"]].head()
```

# save to CSV

```python
df.to_csv('raw_review_netflix.csv', index=False)
print("Data mentah aman yes")
```
