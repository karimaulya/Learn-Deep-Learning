# Import library

- **google_play_scraper**: mengambil data dari play store
- **Sort**: mengurutkan data
- **review**: ulasan
- **pandas**: untuk mengatur dataframe/tabel data

```python
from google_play_scraper import Sort, reviews
import pandas as pd
```

# ID aplikasi Roblox

```python
app_id = 'com.roblox.client'
```

# Bagian yang diambil
- Dari ulasan aplikasi roblox
- Ulasan dengan Bahasa Indonesia
- Negara Indonesia
- Ulasan diurutkan dari yang terbaru
- Target ulasan 10.000 ambil lebih jadi 11.000
- Ulasan dari 1-5 bintang diambil tanpa terkecuali

```python
def scraping_reviews(target):
  result = reviews(
    app_id,
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=target,
    filter_score_with=None
  )

  df = pd.DataFrame(result[0])
  df = df[['userName', 'content', 'score', 'at']]
  return df
```

# Scraping

```python
df_playstore = scraping_reviews(11000)
print(f"Jumlah data yang berhasil diambil: {len(df_playstore)}")
df_playstore.head()
```
